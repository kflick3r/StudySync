# Logistic Regression for Text Classification

Logistic regression is a supervised machine learning algorithm commonly used for classification. Despite its name, it predicts probabilities for categories rather than continuous numerical values.

In NLP, logistic regression can be used for sentiment analysis, spam detection, document classification, and other tasks where text must be assigned to a category.

For binary classification, the model predicts the probability that an input belongs to the positive class. A common decision threshold is 0.5. Predictions with probabilities greater than or equal to 0.5 are assigned to class 1, while lower probabilities are assigned to class 0.

Logistic regression is a **discriminative model** because it directly models the probability of a class given the input features, rather than modeling how the features were generated.

## Text Representation

Machine learning models require numerical input, so text must be converted into feature vectors.

Two common representations are **Bag-of-Words (BoW)** and **TF-IDF**.

Bag-of-Words represents documents using word counts or binary indicators of word presence. It generally ignores word order.

TF-IDF (Term Frequency–Inverse Document Frequency) weights terms according to how frequently they appear in a document and how rare they are across the document collection. Common words that appear in nearly every document receive lower weights.

For example, consider these reviews:

- "The movie was excellent and entertaining."
- "The movie was boring and disappointing."

The vocabulary might include "excellent," "entertaining," "boring," and "disappointing."

Logistic regression learns a weight for each feature. Words associated with positive sentiment may receive positive weights, while words associated with negative sentiment may receive negative weights.

These weights are learned from the training data. Words do not automatically receive positive or negative weights based on their meaning.

## Logistic Regression Model

The model begins by calculating a weighted sum of the input features:

$$z = \mathbf{w}^T\mathbf{x} + b$$

where:

- $\mathbf{x}$ is the input feature vector.
- $\mathbf{w}$ contains the learned feature weights.
- $b$ is the bias term.
- $z$ is the linear score.

The score is passed through the **sigmoid function**:

$$\sigma(z) = \frac{1}{1+e^{-z}}$$

The sigmoid function maps any real-valued number to a value between 0 and 1.

For example:

- $z=0$ gives a probability of $0.5$.
- $z=2$ gives a probability of approximately $0.881$.
- $z=-2$ gives a probability of approximately $0.119$.

The output represents the model's estimated probability of the positive class, although these probabilities are not necessarily well-calibrated.

With a classification threshold of 0.5, the decision boundary occurs when $z=0$.

## Training and Optimization

During training, logistic regression learns the weights and bias that minimize prediction errors.

The model typically uses **binary cross-entropy**, also called log loss:

$$L(y,p) = -[y\log(p)+(1-y)\log(1-p)]$$

Here, $y$ is the true label (0 or 1), and $p$ is the predicted probability of class 1.

Cross-entropy penalizes incorrect predictions, especially when the model is highly confident in an incorrect answer.

For example, predicting $p=0.99$ when the correct label is 0 produces a much larger loss than predicting $p=0.60$ for the same example.

**Gradient descent** is one method for minimizing the loss. It updates model parameters in the direction that reduces the loss:

$$\mathbf{w}\_{t+1} = \mathbf{w}*t - \alpha \nabla*{\mathbf{w}}L $$

The learning rate $\alpha$ controls the size of each update.

A learning rate that is too large may cause unstable optimization, while a learning rate that is too small can make training unnecessarily slow.

Common approaches include batch gradient descent, stochastic gradient descent (SGD), and mini-batch gradient descent. Other numerical optimization algorithms can also be used.

## Regularization

Overfitting occurs when a model learns patterns specific to its training data that do not generalize well to new examples.

This is particularly relevant in NLP because text representations often contain thousands of features, including rare words that appear only a few times.

Regularization discourages unnecessarily large weights and can improve generalization.

**L1 regularization** penalizes the absolute values of the weights:

$$\lambda \sum\_{j=1}^{n}|w_j|$$

L1 can encourage some weights to become exactly zero, effectively performing feature selection.

**L2 regularization** penalizes squared weights:

$$\lambda \sum\_{j=1}^{n}w_j^2$$

L2 encourages smaller weights without generally forcing them to zero.

The regularization strength is a hyperparameter.

In scikit-learn's `LogisticRegression`, the parameter `C` controls the inverse of regularization strength. A smaller `C` means stronger regularization, while a larger `C` means weaker regularization.

## Model Evaluation

Accuracy measures the proportion of correct predictions, but it can be misleading when classes are imbalanced.

For example, if 95% of emails are legitimate, a classifier that always predicts "not spam" would achieve 95% accuracy while failing to identify any spam.

Other important metrics include:

- **Precision:** The proportion of predicted positives that are actually positive.
- **Recall:** The proportion of actual positives correctly identified.
- **F1-score:** The harmonic mean of precision and recall.
- **Confusion matrix:** A table showing true positives, true negatives, false positives, and false negatives.

The formulas are:

$$\text{Precision} = \frac{TP}{TP+FP}$$

$$\text{Recall} = \frac{TP}{TP+FN}$$

$$F_1 = 2\frac{\text{Precision}\cdot\text{Recall}}{\text{Precision}+\text{Recall}}$$

Consider a spam classifier with:

- True positives: 40
- False positives: 10
- False negatives: 20
- True negatives: 130

The resulting metrics are:

- Accuracy: $170/200=85%$
- Precision: $40/50=80%$
- Recall: $40/60\approx66.7%$
- F1-score: approximately $72.7%$

The classifier identifies many spam messages but still misses one-third of the actual spam.

The most appropriate evaluation metric depends on the application. For example, falsely marking legitimate emails as spam may be particularly costly, making precision an important consideration.

## Logistic Regression vs. Naive Bayes

Both logistic regression and Naive Bayes are commonly used as baseline models for text classification.

Naive Bayes is a **generative classifier** that models class probabilities and feature likelihoods. It uses Bayes' theorem and typically assumes that features are conditionally independent given the class.

Logistic regression is **discriminative** and directly estimates the probability of a class given its features.

Naive Bayes can perform well with smaller training datasets because of its simplifying assumptions.

Logistic regression may perform better with sufficient labeled training data, although performance depends on the dataset and feature representation.

Neither algorithm automatically understands the meaning of language. Both rely heavily on the quality of the features and training examples.

## Limitations

Logistic regression is efficient, interpretable, and often provides a strong baseline for text classification. However, it has several limitations.

When using Bag-of-Words or standard TF-IDF features, the model generally ignores word order. For example, "dog bites man" and "man bites dog" contain the same words but have different meanings.

It can also struggle with negation and sarcasm. The sentence "I thought this movie would be great, but it was not" contains the positive word "great," even though the overall sentiment is negative.

Logistic regression learns a linear decision boundary in the chosen feature space. It cannot directly represent complex nonlinear relationships without additional feature engineering or transformations.

Performance also depends on training data quality. If the training examples are unrepresentative or contain systematic biases, the model may perform poorly on new text.

## Improving Performance

Several techniques can improve logistic regression performance in NLP:

- Use consistent text preprocessing, including tokenization and appropriate normalization.
- Compare Bag-of-Words and TF-IDF representations.
- Include word bigrams to capture phrases such as "not good."
- Tune regularization strength using cross-validation.
- Evaluate performance using multiple metrics instead of accuracy alone.
- Inspect incorrectly classified examples to identify recurring errors.
- Keep training and test data separate to avoid data leakage.

Feature extraction should be fitted using only training data. For example, fitting a TF-IDF vectorizer on the entire dataset before splitting can leak information from the test set into the training process.

A typical classification workflow involves collecting labeled examples, splitting the dataset, fitting a vectorizer on the training data, transforming text into numerical features, training the classifier, tuning hyperparameters, and evaluating performance on a held-out test set.

## Key Takeaways

Logistic regression is a supervised, discriminative classification algorithm that estimates class probabilities using a linear combination of input features and the sigmoid function.

In NLP, it is commonly combined with Bag-of-Words or TF-IDF for tasks such as sentiment analysis and spam detection.

Training minimizes cross-entropy loss, while regularization helps prevent overfitting. Performance should be evaluated using appropriate metrics, including precision, recall, and F1-score.

Although logistic regression is efficient and interpretable, its effectiveness depends on feature representation, training data quality, and careful evaluation. It remains an important baseline for comparing more complex NLP models.
